```python
import numpy as np
import matplotlib.pyplot as plt

# Create a grid of values
ccl4 = np.linspace(-10, 10, 50)
ccl3 = np.linspace(-10, 10, 50)
CCL4, CCL3 = np.meshgrid(ccl4, ccl3)

# Define the function
Y = 0.1463 * CCL4 - 0.00411 * CCL3 - 2.98

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(CCL4, CCL3, Y)

# Labels
ax.set_xlabel('CCL4')
ax.set_ylabel('CCL3')
ax.set_zlabel('Y')

plt.show()
```
